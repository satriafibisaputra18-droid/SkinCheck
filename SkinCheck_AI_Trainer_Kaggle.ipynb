# SkinCheck AI - Multi-Task Skin Condition Trainer (Kaggle Edition v2)
Melatih **MobileNetV3-Large** dengan pretrained ImageNet weights untuk 4 indikator visual kulit:
acne_score, redness_score, hyperpigmentation_score, texture_score (skala 0-100).

### Dataset Direkomendasikan (tambahkan via + Add Data):
- **ACNE04** (ruisondias/acne04)
- **Skin Disease Dataset** (subirbiswas19/skin-disease-dataset)
- **HAM10000** (kmader/skin-lesion-analysis-toward-melanoma-detection)

Notebook bisa jalan tanpa dataset eksternal via synthetic fallback.

```python
import subprocess, sys
subprocess.run([sys.executable, '-m', 'pip', 'install', 'onnx', 'onnxruntime', '-q'], check=False)

import os, json, time, shutil, random
from pathlib import Path
import numpy as np, cv2
from PIL import Image
import torch, torch.nn as nn
import torchvision.models as models
import torchvision.transforms as T
from torch.utils.data import Dataset, DataLoader, random_split
from torch.optim.lr_scheduler import CosineAnnealingLR

print(f'PyTorch : {torch.__version__}')
print(f'CUDA    : {torch.cuda.is_available()}')
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f'Device  : {device}')
if torch.cuda.is_available():
    print(f'GPU     : {torch.cuda.get_device_name(0)}')

BASE = Path('/kaggle/working/skincheck')
DS   = BASE / 'dataset'
CKPT = BASE / 'checkpoints'
EXP  = BASE / 'exports'
CATS = ['acne', 'redness', 'hyperpigmentation', 'texture', 'normal_skin']
for c in CATS: (DS / c).mkdir(parents=True, exist_ok=True)
CKPT.mkdir(parents=True, exist_ok=True)
EXP.mkdir(parents=True, exist_ok=True)
print('Dirs ready!')
```

```python
KW = {
  'acne':             ['acne','jerawat','comedo','pimple','papule','pustule','nodule','whitehead','blackhead','blemish','cyst','grade'],
  'redness':          ['redness','erythema','rosacea','merah','kemerahan','rash','eczema','dermatitis','blush','irritat'],
  'hyperpigmentation':['hyperpigment','pigment','melasma','dark_spot','flek','sunspot','freckle','nevus','melanoma','age_spot','lentigo'],
  'texture':          ['texture','wrinkle','keriput','pore','pori','rough','kasar','scar','keratosis','atopic','psoriasis','seborrheic'],
  'normal_skin':      ['normal','clear','healthy','clean','sehat','mulus']
}
valid_ext = {'.jpg','.jpeg','.png','.webp','.bmp'}
inp = Path('/kaggle/input')
imgs = [p for p in inp.rglob('*') if p.suffix.lower() in valid_ext] if inp.exists() else []
print(f'Found {len(imgs)} images in /kaggle/input/')

cnt = {c:0 for c in CATS}
def get_cat(p):
    s = str(p).lower().replace('\\','/')
    for c,kws in KW.items():
        for k in kws:
            if k in s: return c
    return min(cnt, key=cnt.get)

MAX = 500
for p in imgs:
    c = get_cat(p)
    if cnt[c] >= MAX: continue
    try:
        im = cv2.imread(str(p))
        if im is None or im.shape[0]<64 or im.shape[1]<64: continue
        im = cv2.resize(im,(256,256),interpolation=cv2.INTER_AREA)
        cv2.imwrite(str(DS/c/f'{c}_{cnt[c]:05d}.jpg'), im, [int(cv2.IMWRITE_JPEG_QUALITY),92])
        cnt[c]+=1
    except: pass

TONE = {'acne':(100,80,80),'redness':(80,60,80),'hyperpigmentation':(70,75,90),'texture':(110,95,90),'normal_skin':(130,110,100)}
MIN=60
for c in CATS:
    files = list((DS/c).glob('*.jpg'))
    cur = len(files)
    if cur==0:
        print(f'[WARN] {c}: 0 images, generating {MIN} synthetic...')
        for i in range(MIN):
            noise = np.random.randint(-25,25,(256,256,3),dtype=np.int16)
            img = np.clip(np.array(TONE[c],dtype=np.int16)+noise,0,255).astype(np.uint8)
            img = np.broadcast_to(img,(256,256,3)).copy()
            cv2.imwrite(str(DS/c/f'{c}_syn_{i:04d}.jpg'),img)
    elif cur<MIN:
        print(f'[INFO] {c}: {cur} images, augmenting to {MIN}...')
        idx=0
        while cur<MIN:
            im=cv2.imread(str(files[idx%len(files)]))
            if im is not None:
                ops=[lambda x:cv2.flip(x,1),lambda x:cv2.GaussianBlur(x,(5,5),0),
                     lambda x:cv2.convertScaleAbs(x,alpha=1.2,beta=10),lambda x:cv2.rotate(x,cv2.ROTATE_90_CLOCKWISE)]
                aug=ops[idx%4](im)
                cv2.imwrite(str(DS/c/f'{c}_aug_{cur:04d}.jpg'),aug)
                cur+=1
            idx+=1

print('\nFinal counts:')
total=0
for c in CATS:
    n=len(list((DS/c).glob('*.jpg'))); total+=n; print(f'  {c:25s}: {n}')
print(f'  {"TOTAL":25s}: {total}')
```

```python
LABELS = {
  'acne':             [0.88, 0.40, 0.10, 0.55],
  'redness':          [0.10, 0.85, 0.10, 0.15],
  'hyperpigmentation':[0.10, 0.15, 0.88, 0.20],
  'texture':          [0.10, 0.10, 0.15, 0.88],
  'normal_skin':      [0.04, 0.04, 0.04, 0.05],
}
# Order: [acne, redness, hyperpig, texture]

train_tf = T.Compose([
    T.RandomResizedCrop(224, scale=(0.45,1.0), ratio=(0.75,1.33)),
    T.RandomHorizontalFlip(p=0.5), T.RandomVerticalFlip(p=0.1),
    T.RandomRotation(30), T.RandomPerspective(0.35, p=0.5),
    T.ColorJitter(brightness=0.3, contrast=0.3, saturation=0.3, hue=0.05),
    T.RandomGrayscale(p=0.03), T.ToTensor(),
    T.Normalize([0.485,0.456,0.406],[0.229,0.224,0.225]),
    T.RandomErasing(p=0.15, scale=(0.02,0.1))
])
val_tf = T.Compose([
    T.Resize((240,240)), T.CenterCrop(224), T.ToTensor(),
    T.Normalize([0.485,0.456,0.406],[0.229,0.224,0.225])
])

class SkinDS(Dataset):
    def __init__(self, ds_dir, cats, labels, tf=None):
        self.samples, self.tf = [], tf
        for c in cats:
            t = torch.tensor(labels[c], dtype=torch.float32)
            for f in (ds_dir/c).glob('*.jpg'): self.samples.append((f,t))
        random.shuffle(self.samples)
    def __len__(self): return len(self.samples)
    def __getitem__(self, i):
        p,t = self.samples[i]
        im = Image.open(p).convert('RGB')
        return (self.tf(im) if self.tf else T.ToTensor()(im)), t

full = SkinDS(DS, CATS, LABELS, tf=train_tf)
v = max(10, int(0.15*len(full)))
tr = len(full)-v
train_set, val_set = random_split(full,[tr,v],generator=torch.Generator().manual_seed(42))

BATCH=32
tl = DataLoader(train_set, batch_size=BATCH, shuffle=True,  num_workers=2, pin_memory=True)
vl = DataLoader(val_set,   batch_size=BATCH, shuffle=False, num_workers=2, pin_memory=True)
print(f'Train: {len(train_set)} | Val: {len(val_set)} | Batch: {BATCH}')
```

```python
class SkinCheckNet(nn.Module):
    def __init__(self):
        super().__init__()
        base = models.mobilenet_v3_large(weights=models.MobileNet_V3_Large_Weights.IMAGENET1K_V2)
        self.backbone = base.features
        self.pool = nn.AdaptiveAvgPool2d((1,1))
        D = 960
        def head(d):
            return nn.Sequential(
                nn.Linear(d,512), nn.LayerNorm(512), nn.Hardswish(), nn.Dropout(0.4),
                nn.Linear(512,128), nn.ReLU(True), nn.Dropout(0.2),
                nn.Linear(128,1), nn.Sigmoid()
            )
        self.h_acne  = head(D); self.h_red   = head(D)
        self.h_hyper = head(D); self.h_tex   = head(D)
    def forward(self,x):
        f = torch.flatten(self.pool(self.backbone(x)),1)
        return torch.cat([self.h_acne(f)*100, self.h_red(f)*100,
                          self.h_hyper(f)*100, self.h_tex(f)*100], dim=1)

model = SkinCheckNet().to(device)
tp = sum(p.numel() for p in model.parameters())
print(f'Params: {tp:,} | Device: {next(model.parameters()).device}')
```

```python
EPOCHS=25; LR=1e-4; LR_MIN=1e-6; CLIP=1.0
loss_fn   = nn.SmoothL1Loss(beta=5.0)
optim     = torch.optim.AdamW(model.parameters(), lr=LR, weight_decay=1e-3)
sched     = CosineAnnealingLR(optim, T_max=EPOCHS, eta_min=LR_MIN)
best_vl   = float('inf')
best_ckpt = CKPT/'best.pth'
hist      = {'tl':[],'vl':[],'lr':[]}

print(f'Epochs={EPOCHS} | LR={LR} | Scheduler=CosineAnnealing | Loss=SmoothL1')
print('='*65)
t0=time.time()

for ep in range(1, EPOCHS+1):
    model.train(); tl_sum=0.0
    for imgs,tgts in tl:
        imgs=imgs.to(device,non_blocking=True)
        tgts=(tgts*100).to(device,non_blocking=True)
        optim.zero_grad(set_to_none=True)
        loss=loss_fn(model(imgs),tgts)
        loss.backward()
        nn.utils.clip_grad_norm_(model.parameters(),CLIP)
        optim.step()
        tl_sum+=loss.item()*imgs.size(0)
    tl_avg=tl_sum/len(train_set)

    model.eval(); vl_sum=0.0
    with torch.no_grad():
        for imgs,tgts in vl:
            imgs=imgs.to(device,non_blocking=True)
            tgts=(tgts*100).to(device,non_blocking=True)
            vl_sum+=loss_fn(model(imgs),tgts).item()*imgs.size(0)
    vl_avg=vl_sum/len(val_set)

    clr=sched.get_last_lr()[0]; sched.step()
    hist['tl'].append(tl_avg); hist['vl'].append(vl_avg); hist['lr'].append(clr)
    elapsed=time.time()-t0; eta=(elapsed/ep)*(EPOCHS-ep)
    star=' BEST' if vl_avg<best_vl else ''
    print(f'[{ep:02d}/{EPOCHS}] Train={tl_avg:.4f} Val={vl_avg:.4f} LR={clr:.2e} ETA={eta/60:.1f}m{star}')
    if vl_avg<best_vl:
        best_vl=vl_avg
        torch.save({'epoch':ep,'state':model.state_dict(),'val_loss':vl_avg}, best_ckpt)

print('='*65)
print(f'Done! {(time.time()-t0)/60:.1f}min | Best Val Loss: {best_vl:.4f}')
```

```python
try:
    import matplotlib; matplotlib.use('Agg')
    import matplotlib.pyplot as plt
    fig,(a1,a2)=plt.subplots(1,2,figsize=(14,4))
    a1.plot(hist['tl'],label='Train',color='#3b82f6',lw=2)
    a1.plot(hist['vl'],label='Val',  color='#ef4444',lw=2)
    a1.set_title('Loss Curve',fontsize=13,fontweight='bold')
    a1.set_xlabel('Epoch'); a1.set_ylabel('SmoothL1'); a1.legend(); a1.grid(alpha=0.3)
    a2.plot(hist['lr'],color='#10b981',lw=2)
    a2.set_title('LR Schedule',fontsize=13,fontweight='bold')
    a2.set_xlabel('Epoch'); a2.set_yscale('log'); a2.grid(alpha=0.3)
    plt.tight_layout()
    plt.savefig('/kaggle/working/training_curve.png',dpi=120,bbox_inches='tight')
    plt.show(); print('Curve saved.')
except Exception as e: print(f'Plot skipped: {e}')
```

```python
ckpt=torch.load(best_ckpt,map_location='cpu')
model.load_state_dict(ckpt['state'])
model.eval().cpu()
print(f'Loaded best model (epoch={ckpt["epoch"]}, val_loss={ckpt["val_loss"]:.4f})')

dummy=torch.randn(1,3,224,224)

# ONNX
onnx_p=Path('/kaggle/working/skincheck_model.onnx')
torch.onnx.export(model, dummy, str(onnx_p), export_params=True, opset_version=13,
    do_constant_folding=True, input_names=['input_image'], output_names=['skin_indicators'],
    dynamic_axes={'input_image':{0:'batch'},'skin_indicators':{0:'batch'}})
shutil.copy(str(onnx_p), str(EXP/'skincheck_model.onnx'))
print(f'ONNX: {onnx_p.stat().st_size/1e6:.1f} MB')

try:
    import onnx
    onnx.checker.check_model(onnx.load(str(onnx_p)))
    print('ONNX check: OK')
except Exception as e: print(f'ONNX check: {e}')

# TorchScript
pt_p=Path('/kaggle/working/skincheck_model.pt')
torch.jit.trace(model,dummy).save(str(pt_p))
shutil.copy(str(pt_p), str(EXP/'skincheck_model.pt'))
print(f'TorchScript: {pt_p.stat().st_size/1e6:.1f} MB')

# Metadata
meta={'model_name':'SkinCheck-MultiTask-Net','version':'2.0.0',
      'architecture':'MobileNetV3-Large MultiTask (pretrained ImageNet)',
      'epochs_trained':ckpt['epoch'],'best_val_loss':round(float(ckpt['val_loss']),6),
      'input_size':[224,224,3],'input_mean':[0.485,0.456,0.406],'input_std':[0.229,0.224,0.225],
      'output_order':['acne_score','redness_score','hyperpigmentation_score','texture_score'],
      'indicators':[
          {'name':'acne','label':'Jerawat','scale':[0,100]},
          {'name':'redness','label':'Kemerahan','scale':[0,100]},
          {'name':'hyperpigmentation','label':'Hyperpigmentasi','scale':[0,100]},
          {'name':'texture','label':'Tekstur Kulit','scale':[0,100]}
      ],
      'non_medical_notice':'Estimated cosmetic visual indicators only. NOT for medical diagnosis.'}
meta_p=Path('/kaggle/working/model_metadata.json')
json.dump(meta, open(meta_p,'w'), indent=2)
shutil.copy(str(meta_p), str(EXP/'model_metadata.json'))
print('Metadata saved.')

c=Path('/kaggle/working/training_curve.png')
if c.exists(): shutil.copy(str(c), str(EXP/'training_curve.png'))

zip_b='/kaggle/working/skincheck_ai_exported'
shutil.make_archive(zip_b,'zip',EXP)
print(f'ZIP: {zip_b}.zip ({Path(zip_b+".zip").stat().st_size/1e6:.1f} MB)')
print('\n=== SELESAI! Download skincheck_ai_exported.zip dari tab Output ===')
```

```python
import onnxruntime as ort
sess=ort.InferenceSession(str(onnx_p),providers=['CPUExecutionProvider'])
nm=sess.get_inputs()[0].name
print(f'Input: {nm} {sess.get_inputs()[0].shape}')
print(f'Output: {sess.get_outputs()[0].name}')
print()
names=['acne','redness','hyperpig','texture']
for i in range(3):
    d=np.random.randn(1,3,224,224).astype(np.float32)
    raw=sess.run(None,{nm:d})[0][0]
    sc=[int(np.clip(round(float(v)),0,100)) for v in raw]
    pen=sc[0]*.35+sc[1]*.25+sc[2]*.20+(100-sc[3])*.20
    ov=int(np.clip(round(100-pen),0,100))
    print(f'Trial {i+1}: '+' | '.join(f'{n}={v}' for n,v in zip(names,sc))+f' | overall={ov}/100')
print('\nModel ready! Taruh .onnx dan .json ke storage/models/ di project Tuan.')
```
